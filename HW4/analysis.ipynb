2. Explain how hidden missing values (such as ? symbols) impact raw medical datasets, and justify your choice of imputation strategies for skewed numerical vitals versus categorical diagnostic fields.

The hidden missing values can cause errors during training or cause inaccurate predictions. The values should be converted into NaN so they can be identified. With numerical values, median imputation works since medical values can contain outliers or skewed values. While categorical values, mode imputation is better since it replaces the values with the most common category.
