Q3: Why is it critical to fit the PCA transformer strictly on the training split rather than on the entire dataset prior to splitting? What negative impacts occur if PCA is fitted on the full dataset before training downstream classifiers?


It’s important to fit PCA only on the training data to prevent data from leaking. If you were to fit the PCA with the entire dataset, it uses information from the test data before the model is evaluated. This can make the model's accuracy appear higher than it is, it also makes the results less reliable because the test data is no longer completely unseen. Fitting PCA only on the training set provides a fair evaluation of how well the model performs on new data.
