**Q1: Explain how implicit missing values impact medical datasets, and justify your choice of imputation strategy when handling these zeros compared to outright row deletion.**

Missing values happen when medical measurements are 0 instead of being labeled as missing data. An example in the diabetes dataset are measurements like glucose, blood pressure and insulin, which are recorded as 0. For handling those 0s, I used simpleimputer with a median strategy, I used it since the median values are affected less by extreme values. Doing this helps preserve the datasets size and allows for the machine to learn from different patient records. If you were to delete an entire row, this would remove a patient's info which could lead to bias since we won't have the valuable data.

**Q2: Contrast how hard voting aggregates predictions versus soft voting. Why do well-calibrated probability estimates matter when implementing soft voting across heterogeneous models?**

The difference between the two comes down to the prediction. Hard voting combines different predictions from multiple classifiers, selecting the class that receives a majority of the votes. While soft voting provides better predictions since it considers how confident each model is.  Though, probability estimates must be calibrated so that the model's confidence is accurate and correct. Without properly calibrating, an overconfident model can affect the final prediction too heavily and be inaccurate or incorrect.

**Q3: What is the role of out-of-bag evaluation in Bagging classifiers, and why does an OOB score serve as an efficient alternative to traditional cross-validation splits during ensemble training?**

The role for out of bag is it evaluates measurements on how well a bagging classifier performed on the training observations that werent added in individual samples. Since each of the trees train on random samples, about 37% of the training observations are left out. The observations that were included can then be used to evaluate predictions without requiring a different dataset. Making out of bag more efficient than a traditional cross validation since additional models dont need to be trained.

**Q4: How does introducing random feature subsetting in Random Forests differ from standard Bagging, and how does this feature diversity impact model variance and generalization?**

With standard bagging, it trains multiple trees using random bootstrap samples, each tree can consider all the features that are available when splitting. With random forests, it selects a subset of features at each split. What this does is that it prevents the trees from relying on the same dominant variable and makes the models train and learn different patterns. This makes the trees different from one another, reducing errors. Random forests can be more accurate and improve generalization when predicting diabetes.

**Q5: Discuss the mechanics of sequential sample weighting in AdaBoost. Based on your training versus test accuracy results, did AdaBoost exhibit signs of overfitting, and how can hyperparameter adjustments mitigate this?**

What adaboost does is it trains weak learners like decision stumps, it adjusts sample weight after each training. Misclassified values receive a higher weight, making the models focus heavily on difficult predictions. When testing accuracy becomes higher, overfitting can occur, this indicates that the model has learned patterns that don't generalize the outcomes. Reducing the values of estimators, learning rate or limiting tree patterns can help control overfitting.
